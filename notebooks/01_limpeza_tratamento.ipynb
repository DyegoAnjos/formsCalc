```python
import pandas as pd
```

```python
df = pd.read_csv('../data/raw/curso_de_feria_matematica.csv')
```

```python
print("Colunas:")
for coluna in df.columns:
    print(f"'{coluna}'")
```

```python
dicionario_original = dict(enumerate(df.columns))
```

```python
mapa_colunas={
    'Carimbo de data/hora': 'data_hora',
    'Endereço de e-mail': 'e_mail',
    'Nome completo:': 'nome',
    'Matrícula:': 'matricula',
    'Período atual:': 'periodo',
    'Você teria interesse em cursar alguma disciplina do eixo de matemática durante as férias? ': 'interesse',
    'Qual é a sua situação em cada disciplina?  [Fundamentos de Cálculo (TMT0043)]': 'status_fcal',
    'Qual é a sua situação em cada disciplina?  [Álgebra Linear (TMT0044)]': 'status_linear',
    'Qual é a sua situação em cada disciplina?  [Cálculo Diferencial e Integral I (TMT0045)]': 'status_calc1',
    'Qual é a sua situação em cada disciplina?  [Cálculo Diferencial e Integral II (TMT0046)]': 'status_calc2',
    'Em quais dias e turnos você poderia frequentar as aulas?  [Segunda-feira]': 'turno_segunda',
    'Em quais dias e turnos você poderia frequentar as aulas?  [Terça-feira]': 'turno_terca',
    'Em quais dias e turnos você poderia frequentar as aulas?  [Quarta-feira]': 'turno_quarta',
    'Em quais dias e turnos você poderia frequentar as aulas?  [Quinta-feira]': 'turno_quinta',
    'Em quais dias e turnos você poderia frequentar as aulas?  [Sexta-feira]': 'turno_sexta',
}
```

```python
df = df.rename(columns=mapa_colunas)
```

```python
print("Novas Colunas:")
print(df.columns.tolist())
```

```python
df.info()
```

```python
if 'data_hora' in df.columns:
    df['data_hora'] = pd.to_datetime(df['data_hora'], dayfirst=True)

df['data_envio'] = df['data_hora'].dt.date
df['dia_semana'] = df['data_hora'].dt.day_name(locale='pt_BR.utf8')
df['hora_envio'] = df['data_hora'].dt.hour
```

```python
colunas_turno = ['turno_segunda','turno_terca','turno_quarta','turno_quinta','turno_sexta']
for coluna in colunas_turno:
    if coluna in df.columns:
        df[coluna] = df[coluna].fillna('Sem resposta')
```

```python
nulos_por_coluna = df.isna().sum()
colunas_com_nan = nulos_por_coluna[nulos_por_coluna > 0]

print(colunas_com_nan)
```
